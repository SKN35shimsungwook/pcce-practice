## 111
사용자로부터 입력받은 문자열을 두 번 출력하라.

```python
# 사용자로부터 문자열을 입력받는다
user = input("입력: ")

# 문자열에 * 2를 하면 문자열이 그대로 두 번 이어붙여진다
print(user * 2)
```

## 112
사용자로부터 하나의 숫자를 입력받고, 입력받은 숫자에 10을 더해 출력하라.

```python
# input()의 결과는 항상 문자열(str)이므로 int()로 형변환 필요
user = input("숫자를 입력하세요: ")

# 문자열 -> 정수 변환 후 10을 더해 출력
print(10 + int(user))
```

## 113
사용자로부터 하나의 숫자를 입력 받고 짝수/홀수를 판별하라.

```python
user = input("")

# 정수로 변환한 뒤 2로 나눈 나머지가 0이면 짝수
if int(user) % 2 == 0:
    print("짝수")
else:
    print("홀수")
```

## 114
사용자로부터 값을 입력받아 20을 더한 값을 출력하되, 255를 초과하면 255를 출력하라.

```python
user = input("입력값: ")
num = 20 + int(user)

# 계산 결과가 255를 초과하면 255로 고정(클리핑)
if num > 255:
    print(255)
else:
    print(num)
```

## 115
사용자로부터 값을 입력받아 20을 뺀 값을 출력하되, 결과 범위는 0~255로 제한하라.

```python
user = input("입력값: ")
num = int(user) - 20

# 상한(255)과 하한(0)을 각각 체크하여 범위를 벗어나지 않도록 클리핑
if num > 255:
    print(255)
elif num < 0:
    print(0)
else:
    print(num)
```

## 116
사용자로부터 입력받은 시간이 정각인지 판별하라. (예: 02:00)

```python
time = input("현재시간: ")

# 문자열 슬라이싱으로 뒤에서 2자리(분)를 추출해 "00"인지 비교
if time[-2:] == "00":
    print("정각 입니다.")
else:
    print("정각이 아닙니다.")
```

## 117
사용자가 입력한 단어가 fruit 리스트에 포함되어 있는지 확인하라.

```python
fruit = ["사과", "포도", "홍시"]
user = input("좋아하는 과일은? ")

# in 연산자로 리스트에 포함 여부 확인
if user in fruit:
    print("정답입니다.")
else:
    print("오답입니다.")
```

## 118
투자 경고 종목 리스트에 입력한 종목명이 포함되어 있는지 확인하라.

```python
warn_investment_list = ["Microsoft", "Google", "Naver", "Kakao", "SAMSUNG", "LG"]
stock = input("종목명: ")

# in 연산자로 경고 종목 리스트 포함 여부 확인
if stock in warn_investment_list:
    print("투자 경고 종목입니다.")
else:
    print("투자 경고 종목이 아닙니다.")
```

## 119
fruit 딕셔너리의 키(key) 값에 사용자 입력이 포함되어 있는지 확인하라.

```python
fruit = {"봄": "딸기", "여름": "토마토", "가을": "사과"}
user = input("제가 좋아하는 계절은: ")

# 딕셔너리에 in을 쓰면 기본적으로 key 값을 검사 대상으로 한다
if user in fruit:
    print("정답입니다.")
else:
    print("오답입니다.")
```

## 120
fruit 딕셔너리의 값(value)에 사용자 입력이 포함되어 있는지 확인하라.

```python
fruit = {"봄": "딸기", "여름": "토마토", "가을": "사과"}
user = input("좋아하는 과일은? ")

# .values()로 value 목록만 뽑아서 in으로 포함 여부 확인
if user in fruit.values():
    print("정답입니다.")
else:
    print("오답입니다.")
```
